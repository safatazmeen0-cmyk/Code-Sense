# CodeSense Preprocessing & Feature Extraction

This notebook demonstrates text cleaning, code token normalization, and feature assembly for the ML classifier.

```python
import re
import pandas as pd

def clean_text(value):
    text = "" if value is None else str(value)
    text = text.lower()
    text = re.sub(r"[^a-z0-9_+*/%=<>()\[\]{}.:,'\"\s\-]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

sample = "Print(  numbers[5]  )!!"
print("Raw sample:", sample)
print("Cleaned sample:", clean_text(sample))
```

### Combining Multi-Field Features

```python
def combine_features(row):
    fields = ["code", "error_message", "error_type", "concept", "keywords"]
    return clean_text(" ".join(str(row.get(field, "")) for field in fields))

df = pd.read_csv("../dataset/error_dataset.csv").fillna("")
df["features"] = df.apply(combine_features, axis=1)
print("Sample engineered feature string:")
print(df["features"].iloc[0])
```

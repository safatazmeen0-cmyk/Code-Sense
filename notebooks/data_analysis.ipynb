# CodeSense Data Analysis

This notebook explores and analyzes the `dataset/error_dataset.csv` used for error classification in CodeSense.

```python
import pandas as pd
from pathlib import Path

DATASET_PATH = Path("../dataset/error_dataset.csv")
df = pd.read_csv(DATASET_PATH)
print("Dataset Shape:", df.shape)
df.head()
```

### Error Category Distribution

```python
print("Category breakdown:")
df["error_category"].value_counts()
```

### Difficulty & Concepts Overview

```python
print("Difficulty breakdown:")
print(df["difficulty"].value_counts())

print("\nTop concepts represented:")
print(df["concept"].value_counts().head(10))
```
